## Обратный код

```python
def to_ones_complement_math(number, bits=8):
    if number >= 0:
        return f"{number:0{bits}b}"
    else:
        # Побитовое НЕ в Python дает (-number - 1), маскируем под разрядность
        mask = (1 << bits) - 1
        return f"{(~abs(number)) & mask:0{bits}b}"

print("Число -5 (8 бит):", to_ones_complement_math(-5, 8))
```

## Дополнительный код

```python
def to_twos_complement(number, bits=8):
    # Определяем границы диапазона для дополнительного кода
    # Например, для 8 бит: от -128 до +127
    min_val = -(1 << (bits - 1))
    max_val = (1 << (bits - 1)) - 1

    if not (min_val <= number <= max_val):
        return f"Ошибка: число {number} не входит в диапазон от {min_val} до {max_val} для {bits} бит"

    # Применяем побитовую маску, чтобы ограничить число нужным количеством бит
    mask = (1 << bits) - 1
    twos_complement_int = number & mask

    # Форматируем в двоичную строку с ведущими нулями
    return f"{twos_complement_int:0{bits}b}"

# --- Примеры использования ---
print("Число  5 (8 бит):", to_twos_complement(5, 8))
print("Число -5 (8 бит):", to_twos_complement(-5, 8))
print("Число -1 (8 бит):", to_twos_complement(-1, 8))
print("Число -128 (8 бит):", to_twos_complement(-128, 8))
print("Число -127 (8 бит):", to_twos_complement(-127, 8))
print("Число -001 (8 бит):", to_twos_complement(-1, 8))
```
