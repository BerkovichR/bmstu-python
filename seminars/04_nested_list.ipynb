## Операции со списком

Метод через точку меняет **этот** список. Сложение, умножение и срез собирают **новый**.

| Операция | Что делает |
|---|---|
| `len(a)` | длина |
| `a[i]`, `a[i] = x` | прочитать и заменить элемент |
| `a[start:end:step]` | срез, новый список |
| `a.append(x)` | добавить `x` в конец |
| `a.pop()`, `a.pop(i)` | снять последний или элемент с индексом `i` |
| `a.insert(i, x)` | вставить `x` на место `i` |
| `a.remove(x)` | убрать первое вхождение значения `x` |
| `a.sort()` | упорядочить этот список |
| `a.clear()` | обнулить длину, шапка та же |
| `a.reverse()` | перевернуть этот список |
| `a.extend(other)` | дописать элементы `other` в конец, без лишней вложенности |
| `a + b` | новый список, склейка |
| `a * n` | новый список, `n` повторов |
| `a.index(x)` | индекс первого `x` |
| `a.count(x)` | сколько раз встретился `x` |
| `a.copy()`, `a[:]` | новая шапка, те же элементы |
| `del a[i]` | удалить элемент с индексом `i` |


## Вложенный список

Элемент списка сам может быть списком. Для матрицы это естественная запись: внешний список — строки, внутренний — элементы строки.

Индекс пишется дважды. `a[i]` — строка номер `i`. `a[i][j]` — элемент этой строки в столбце `j`. Нумерация с нуля, как у плоского списка.


````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
print(a)
print(a[0][2])
print(a[2][0])
````

````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
a[1][2] = 60
print(a)
````

````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
row = a[1]
print(row)
````

## Одна и та же строка на трёх местах

`row` — имя списка. Если три ячейки внешней шапки смотрят на этот один список, замена `m[0][0]` видна во всех трёх строках.

`[[0, 0, 0]] * 3` делает то же самое: внешняя шапка получает три ссылки на одну строку. Повтор числа безопасен (`[0] * 3` — три числа). Повтор списка размножает ссылку, а не копию.


````python
row = [0, 0, 0]
m = [row, row, row]
m[0][0] = 1
print(m)
````

````python
m = [[0, 0, 0]] * 3
m[0][0] = 1
print(m)
````

````python
m = [[0, 0, 0], [0, 0, 0], [0, 0, 0]]
m[0][0] = 1
print(m)
````

`append` кладёт в конец **один** элемент. Если этот элемент — список, появляется новая строка.

`extend` забирает элементы из переданного списка и дописывает их по одному. Вложенности не возникает.


````python
m = []
m.append([1, 2, 3])
print("после append:", m)

n = []
n.extend([1, 2, 3])
print("после extend:", n)
````

## Сумма всех элементов

Накопитель `s` начинает с нуля. Внешний цикл идёт по строкам, внутренний — по элементам строки. Тело цикла пишется со следующей строки.


````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
s = 0
for row in a:
    for x in row:
        s = s + x
print(s)
````

## Генератор списка

Та же сборка нового списка, что цикл с `append`, записанная в одних скобках.

```
[выражение for переменная in список]
```

Слева — что кладём в новый список. Справа — откуда берём.


````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
row = a[0]
squares = []
for x in row:
    squares.append(x * x)
print(squares)
````

````python
row = [1, 2, 3]
squares = [x * x for x in row]
print(squares)
````

Матрица квадратов — тот же генератор строки, и так для каждой строки.

````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
b = []
for row in a:
    new_row = []
    for x in row:
        new_row.append(x * x)
    b.append(new_row)
print(b)
````

````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
b = [[x * x for x in row] for row in a]
print(b)
````

````python
# Геометрическая прогрессия
b, q, n = 3, 2, 10
a = [b * q**i for i in range(n)]
print(a)
````

## Продвинутый синтаксис

В генератор можно добавить условие. В новый список попадает только то, что условию удовлетворяет.

```
[выражение for переменная in список if условие]
```


````python
row = [1, 2, 3]
evens = [x for x in row if x % 2 == 0]
print(evens)
````

Два цикла в одних скобках выписывают матрицу в один список. Сначала внешний цикл по строкам, затем внутренний по элементам строки. Порядок `for` тот же, что у вложенного цикла выше.

```
[выражение for строка in матрица for элемент in строка]
```

Что получится на нашей матрице?


````python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
flat = [x for row in a for x in row]
print(flat)
````

## Кортеж

Кортеж записывается круглыми скобками. Индекс работает так же, как у списка. Заменить элемент нельзя: набор фиксирован.

Кортеж уместен, когда состав не должен меняться по ходу программы: пара координат, два значения, которые идут вместе. Пока набор пополняют и правят, это список.


````python
t = (10, 20, 30)
print(t)
print(t[0])
print(t[2])
````

````python
t = (10, 20, 30)
t[0] = 1
````

````python
a = (1)
b = (1,)
print(type(a), a)
print(type(b), b)
````

## Словарь

У списка место элемента — это индекс `0, 1, 2, ...`. У словаря место элемента — это **ключ**, который задаём сами. Ключ не обязан быть позицией.

`d[ключ]` читает и записывает значение. `del d[ключ]` убирает пару. `d.get(ключ)` на отсутствующем ключе даёт `None` и программу не останавливает. `d.keys()`, `d.values()` и `d.items()` отдают ключи, значения и пары.


````python
d = {"x": 2, "y": -1}
print(d["x"])
d["x"] = 5
print(d)
del d["y"]
print(d)
print(d.get("z"))
print(d.get("x"))
print(list(d.keys()))
print(list(d.values()))
for key, value in d.items():
    print(key, value)
````

## Краткий итог

Вложенный список: `a[i][j]`. Строка `a[i]` — отдельный список. Три ссылки на одну строку меняются вместе. `append` вкладывает, `extend` выпрямляет.

Генератор повторяет цикл с `append`: `[выражение for переменная in список]`.

Кортеж не меняют. Одна запятая в `(1,)` делает кортеж.

Словарь ищут по ключу, список — по индексу.
