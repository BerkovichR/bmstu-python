## Область видимости

Каждый пример с `x` начинается заново с `x = 2`.


Функция читает внешний `x`.


```python
x = 2

def read_x():
    return x

print(read_x())
```

Присваивание внутри создаёт локальный `x`. Внешний `x` остаётся прежним.


```python
x = 2

def write_local():
    x = 3
    return x

print(write_local())
print(x)
```

`global` пишет во внешний `x`.


```python
x = 2

def set_global():
    global x
    x = 3

set_global()
print(x)
```

`nonlocal` пишет в `n` внешней функции.


```python
def outer():
    n = 0
    def inner():
        nonlocal n
        n = n + 1
        return n
    return inner() + inner()

print(outer())
```

## `is_palindrome`

Функция `is_palindrome(word)` возвращает, читается ли слово одинаково с начала и с конца. Регистр тот же. Один конец схемы, раннего выхода нет.

- `is_palindrome('казак')` даёт `True`
- `is_palindrome('дом')` даёт `False`
- `is_palindrome('а')` даёт `True`
- `is_palindrome('')` даёт `True`




`words(text)` заменяет каждый знак из `string.punctuation` пробелом и делит строку через `split`. В `string.punctuation` знаки ASCII. Русских кавычек и тире там нет.

`'казак, топот дом!'` даёт `['казак', 'топот', 'дом']`.

`is_palindrome('Казак')` даёт `False`, `is_palindrome('Казак', ignore_case=True)` даёт `True`.
