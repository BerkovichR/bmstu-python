Присваивание внутри создаёт локальный `x`. Внешний `x` остаётся прежним.


```python
x = 2

def write_local():
    x = 3
    return x

print(write_local())
print(x)
```

`global` пишет во внешний `x`.


```python
x = 2

def set_global():
    global x
    x = 3

set_global()
print(x)
```

`nonlocal` пишет в `n` внешней функции.


```python
def outer():
    n = 0
    def inner():
        nonlocal n
        n = n + 1
        return n
    return inner() + inner()

print(outer())
```

Пример хорошего оформления функции:

```python
def total_shopping_cost(shopping_list):
    """
    Вычисляет общую сумму покупок и их количество.

    Args:
        shopping_list (list): список кортежей или словарей вида [('хлеб', 50), ('молоко', 80)]
                           или [{'name': 'хлеб', 'price': 50}, {'name': 'молоко', 'price': 80}]

    Returns:
        tuple: (общая_сумма, количество_покупок)
    """
    if not shopping_list:
        return 0.0, 0

    total_sum = 0.0
    count = 0

    for item in shopping_list:
        # Формат: ('название', цена) или ('название', цена, количество)
        price = float(item[1])
        total_sum += price
        count += 1

    return total_sum, count
```

## `is_palindrome`

Функция `is_palindrome(word)` возвращает, читается ли слово одинаково с начала и с конца. Регистр тот же. Один конец схемы, раннего выхода нет.

- `is_palindrome('казак')` даёт `True`
- `is_palindrome('дом')` даёт `False`
- `is_palindrome('а')` даёт `True`
- `is_palindrome('')` даёт `True`




`words(text)` заменяет каждый знак из `string.punctuation` пробелом и делит строку через `split`. В `string.punctuation` знаки ASCII. Русских кавычек и тире там нет.

`'казак, топот дом!'` даёт `['казак', 'топот', 'дом']`.

`is_palindrome('Казак')` даёт `False`, `is_palindrome('Казак', ignore_case=True)` даёт `True`.


```python
import string
def words(text:str) -> list:
    s = ""
    for ch in text:
        if ch in string.punctuation:
            s += ' '
        else:
            s += ch
    return s.split()

text = "казак,кaзак          сила"
words(text)
```

```python
def is_palindrom(word:str) -> bool:
    word = word.lower()
    n = len(word)
    # print(n)
    for i in range(n//2):
        # print(i)
        if word[i] != word[n-1-i]:
            # print(ord(word[i]), ord(word[n-1-i]))
            return False
    return True

print(is_palindrom("Казак"))
```

```python
def summary(words:list) -> tuple[int, int]:
    n = len(words)
    cnt = 0
    for word in words:
        if is_palindrom(word):
            cnt += 1
    return n, cnt

summary(["казак","кaзак", "сила"])
```

```python
text = "казак,кaзак          сила"
summary(words(text))

w = words(text)
summary(w)
```
