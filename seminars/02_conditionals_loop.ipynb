### Тернарный оператор

Это **выражение**, не замена `if` / `elif` / `else`. Одна строка: «значение, если условие, иначе другое».


````python
x = 7
print(x if x % 2 == 0 else -x)

# Это эквивалентно
if x % 2 == 0:
    print(x)
else:
    print(-x)
````

### Циклы

Общий синтаксис:
```python
for переменная in коллекция:
    действие с переменная
```

Рассмотрим основные варианты. 
- **`for` + `range`** — число шагов известно заранее.
- **`while`** — заранее неизвестно, сколько раз тело выполнится.

Остальные потом.

### Функция `range`

- `range(n)` — числа `0, 1, …, n-1`
- `range(a, b)` — от `a` до `b`, **конец не входит**
- `range(a, b, step)` — с шагом; шаг может быть отрицательным

`range` — не список. Не превращаем его в список «на всякий случай».


````python
for i in range(5):
    print(i)
````

````python
for i in range(1, 6):
    print(i)
````

````python
for i in range(0, 10, 2):
    print(i)
````

````python
for i in range(5, 0, -1):
    print(i)
````

### `continue` и `break`

Другие способы управления циклом. 

- `continue` — перейти к следующей итерации
- `break` — выйти из цикла

Работает в любых циклах: и `while` и `for` + `range`, что в программе B.


````python
n = int(input())
for i in range(1, n + 1):
    if i % 2 == 0:
        continue
    print(i)
````

````python
s = 0
for i in range(1, 100):
    s += i
    if s > 20:
        break
print(s)
````
