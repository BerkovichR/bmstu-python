## Функции

| Имя | Фраза | Вызов |
|---|---|---|
| `lambda` | функция из одного выражения: слева параметры, справа результат | `lambda x: x * x` |
| `map` | применяет функцию к каждому элементу и отдаёт результаты по одному | `list(map(lambda x: x * x, [1, 2, 3]))` |
| `filter` | оставляет элементы, для которых функция вернула истину | `list(filter(lambda x: x > 0, [1, -2, 3]))` |
| `reduce` | сворачивает список в одно значение слева направо | `reduce(lambda x, y: x + y, [1, 2, 3], 0)` |
| `all` | истина, только если условие выполнено для каждого элемента | `all(x > 0 for x in [1, -2, 3])` |
| `any` | истина, если условие выполнено хотя бы для одного элемента | `any(x > 0 for x in [1, -2, 3])` |
| `accumulate` | каждое значение — итог от начала списка до этого места | `list(accumulate([1, 2, 3, 4]))` |
| `count` | генерирует бесконечную последовательность чисел с заданным шагом | `next(count(start=10, step=2))` |
| `islice` | нарезает любой итератор по индексам, не загружая его целиком в память | `list(islice(count(1), 0, 3))` |



```python
square = lambda x: x * x
print(square(3))
```

```python
print(list(map(lambda x: x * x, [1, 2, 3])))
print(map(lambda x: x * x, [1, 2, 3]))
```

```python
print(list(filter(lambda x: x > 0, [1, -2, 3])))
```

```python
from functools import reduce

print(reduce(lambda x, y: x + y, [1, 2, 3], 10))
```

```python
nums = [1, -2, 3]
print(all(x > 0 for x in nums))
print(any(x > 0 for x in nums))
```

```python
nums = [1, -2, 3]
print(list(map(lambda x: x * x, filter(lambda x: x > 0, nums))))
```

## Функциональный пайплайн анализа IoT-датчиков

Вы разрабатываете модуль аналитики для системы умного дома. Датчики присылают непрерывный поток значений температуры. Вам нужно написать пайплайн обработки этих данных, используя исключительно функциональный подход (без явных циклов for/while и списковых включений там, где можно применить map/filter).


1 Генерация данных (count, map): С помощью itertools.count и map сгенерируйте бесконечную последовательность "сырых" значений, где i-й элемент вычисляется по формуле: value = (i * 7 % 20) - 10.

2 Окно наблюдения: возьмите из этого потока только значения с 15-го по 80-й (индексы 15 до 79 включительно).

3 Очистка: отфильтруйте поток, оставив только значения в диапазоне (-5, 15) (строго между).

4 Трансформация: переведите отфильтрованные значения в "индекс нагрузки" по формуле: abs(x - 5) * 2 + 1 (прибавляем 1, чтобы избежать нулей).

5 Накопление: С помощью itertools.accumulate вычислите нарастающую сумму индексов нагрузки.

6 Валидация состояния (any, all):
- С помощью any проверьте, превысила ли нарастающая сумма когда-либо значение 300.
- С помощью all проверьте, что все значения исходного отфильтрованного списка (до накопления) были строго больше -10.

7 Индексация: пронумеруйте итоговые шаги нарастающей суммы, создав словарь вида {номер_шага: значение_суммы}, используя itertools.count (или collections.Counter для подсчета частот).

```python
from itertools import count, islice, accumulate
from functools import reduce

# 1. Генерация бесконечного потока
raw_stream = map(lambda i: (i * 7 % 20) - 10, count(1))

# 2. Берем окно с 15 по 79 индекс
sliced_stream = islice(raw_stream,15,80)

#3.
filter_num = list(filter(lambda x: -5<x<15, sliced_stream))

# 4.
map_list = map(lambda x: abs(x-5)*2+1, filter_num )

acc_num = accumulate(map_list)

# Вывод результатов
print(f"Данные: {list(acc_num)}")

# print(f"Данных после фильтрации: {len(filtered_data)}")
# print(f"Система перегружалась (>300): {is_overloaded}")
# print(f"Базовые данные валидны: {is_valid_baseline}")
# print(f"Первые 5 шагов отчета: {dict(list(final_report.items())[:5])}")
```
