### `Объекты и переменные`

```python
a = 5
b = a
```

- создан объект типа `int`, в котором содержится значение 5
- затем создана переменная с именем `a`, которая ссылается на этот объект
- создана переменная `b`, которая указывает на тот же объект типа `int`

```python
a = a + 1  # equals to a += 1
print(b)
```

- выражение `a = a + 1` создаёт новый объект типа `int`, равный 6
- переменная `a` начинает ссылаться на него и больше не связана с объектом, равным 5
- `b` продолжает ссылаться на него

### `Объекты и переменные в Python`

Библиотека math. Доступна после `import math`.
Полезные функции.

- `math.floor(x)`: Округляет число вниз до ближайшего целого (к «полу»).
- `math.ceil(x)`: Округление вверх до ближайшего целого (к «потолку»).Пример: math.ceil(4.1) вернет 5.
- `math.fabs(x)`: Возвращает модуль числа (абсолютное значение) в виде числа с плавающей точкой (float). В отличие от встроенной abs(), всегда возвращает вещественное число.
- `math.sqrt(x)`: Вычисляет квадратный корень числа.
- `math.pow(x, y)`: Возводит x в степень y. Всегда возвращает float.
- `math.log(x)`: Вычисляет натуральный (по основанию e).
- `math.log(x, [base])`: Вычисляет логарифм. По умолчанию — натуральный (по основанию e).
- `math.log10(x)`: Десятичный логарифм (по основанию 10).
- `math.sin(x)` / `math.cos(x)` / `math.tan(x)`: Синус, косинус и тангенс угла.
- `math.radians()` Перевод градусов в радианы
- `math.isclose(x, y)` Сравнивает x и y с учетом погрешности/
- math.pi Константа пи
- math.e Константа e

Форматирование строк

В Python есть несколько похожих способов форматирования - выражение и метод

```python
print('x = %d, y = %f' % (10.5, 11))
```

```python
print('x = %o, y = %E' % (8, 1.0 / 3))  # octal format, exponential + upper case
```

```python
print('x = %10.2f' % (1.0 / 3))  # min width and precision
```

Другой способ

```python
print('x = {}, y = {}'.format(10, 20))
```

```python
print('x = {1}, y = {0}'.format(10, 20))
```

Современный способ: f-строки (начиная с Python 3.6)

```python
a=8
 print(f'x = {10}, y = {20}')
```

```python
print(f'x = {2**4}, y = {math.sin(20)}')
```

### `Проверка стиля кода`

- Стандартным набором правил codestyle для Python является PEP8
- Есть несколько инструментов для проверки кода на соответствию стиля: `pep8`, `pyflakes`, `flake8`, `Pylint`
- В рамках курса в качестве линтера используется `flake8`
- Установка модуля стандартная через `pip`, используется он как утилита с управляющими флагами:

```python
import os, shutil
os.mkdir('tmp')
with open('temp.py', 'w') as fout:
    fout.write('\n'.join([
        'print(1111111111111111111111111111111111111111111111111111111111111111111)',
        'print(1111111111111111111111111111111111111111111111111111111111111111111)  # noqa: E501\n',
    ]))

with open(os.path.join('tmp', 'temp.py'), 'w') as fout:
    fout.write('''print(1111111111111111111111111111111111111111111111111111111111111111111)''')
```

```python
!flake8 . --extend-exclude=tmp --show-source --statistics --count --max-line-length=20
shutil.rmtree('tmp')
```
