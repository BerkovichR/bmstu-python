# Занятие 5: словари, множества и строки


## Словарь

Список хранит порядок и даёт элемент по номеру. Словарь хранит пары «ключ — значение» и даёт значение по ключу.


```python
d = {"x": 2, "y": -1}
print(d)
```

`d["x"]` читает значение по ключу `x`.


```python
print(d["x"])
```

Присваивание меняет значение при том же ключе.


```python
d["x"] = 5
print(d)
```

`del` удаляет пару.


```python
del d["y"]
print(d)
```

`get` возвращает значение по ключу. Если ключа нет, `get` даёт `None`, и программа продолжает работу.


```python
print(d.get("z"))
print(d.get("x"))
```

Квадратные скобки на отсутствующем ключе останавливают программу.


```python
d["z"]
```

Ключ не может быть списком.


```python
d[[1, 2]] = 0
```

## Частоты

Список чисел уже задан. Цикл считает, сколько раз встретилось каждое число. `get(x, 0)` даёт 0, если этого ключа ещё нет. Тело цикла пишется со следующей строки.


```python
a = [1, 2, 1, 3, 2, 1]
d = {}
for x in a:
    d[x] = d.get(x, 0) + 1
for key, value in d.items():
    print(key, value)
```

Порядок печати — порядок первого появления числа в списке.

`keys` даёт ключи, `values` даёт значения.


```python
print(list(d.keys()))
print(list(d.values()))
```

## Множество

Множество хранит каждое значение один раз. Номера у элементов нет.

Повтор в литерале исчезает.


```python
s = {1, 2, 2, 3}
print(s)
```

`add` добавляет элемент. `remove` удаляет элемент.


```python
s.add(4)
print(s)
```

```python
s.remove(2)
print(s)
```

Пустые фигурные скобки — словарь. Пустое множество пишется `set()`.


```python
print(type({}))
print(set())
```

`|` — объединение. `&` — пересечение. `-` — разность.


```python
A = {1, 2, 3}
B = {3, 4, 5}
print(A | B)
```

```python
print(A & B)
```

```python
print(A - B)
```

`set` от списка оставляет каждое число один раз.


```python
print(set(a))
```

В множество кладут такие же элементы, как ключи словаря. Список в множество не кладут.


## Краткий итог

Словарь хранит пары. `get` на отсутствующем ключе даёт `None`. Частоты копят циклом.

Множество хранит каждое значение один раз. `|` объединяет, `&` оставляет общее, `-` убирает чужое. `set(список)` снимает дубликаты.

**Словарь**

| Операция | Что делает |
|---|---|
| `len(d)` | число пар |
| `d[key]` | прочитать значение по ключу |
| `d[key] = x` | записать значение; новый ключ добавляет пару |
| `del d[key]` | удалить пару |
| `d.get(key)` | значение или `None`, если ключа нет |
| `d.get(key, x)` | значение или `x`, если ключа нет |
| `d.keys()` | ключи |
| `d.values()` | значения |
| `d.items()` | пары ключ-значение |
| `dict.fromkeys(keys)` | ключи из списка, значения `None` |

**Множество**

| Операция | Что делает |
|---|---|
| `len(s)` | число элементов |
| `{...}` | множество, повторы исчезают |
| `set()` | пустое множество |
| `{}` | пустой словарь, не множество |
| `set(a)` | множество из списка, без повторов |
| `s.add(x)` | добавить элемент |
| `s.remove(x)` | удалить элемент |
| `A \| B` | объединение |
| `A & B` | пересечение |
| `A - B` | разность: есть в `A`, нет в `B` |
| `A ^ B` | симметричная разность |
| `A <= B` | левое множество лежит в правом |
| `A.update(B)` | дописать элементы `B` в это множество |


### Справочно


```python
dict.fromkeys(['a', 'b'])
A ^ B
{1, 3} <= A
A.update(B)
```

## Строки

Строка — последовательность символов в кавычках.


```python
s = 'qwerty'
print(s)
```

Нумерация с нуля. Отрицательный номер считает с конца.


```python
print(s[0])
print(s[-1])
```

Срез собирает новую строку. Правый конец в срез не входит.


```python
print(s[0:3])
```

`len` даёт длину.


```python
print(len(s))
```

Сложение собирает новую строку.


```python
print(s + '!')
```

Умножение на число повторяет строку и собирает новую.


```python
print(s * 3)
```

Замена символа по индексу не выполняется.


```python
s[0] = 'a'
```

`str` делает строку из числа. `int` делает число из строки.


```python
print(str(10), type(str(10)))
print(int('10'), type(int('10')))
```

`format` подставляет значения в фигурные скобки.


```python
print('x = {}'.format(2))
```

f-строка подставляет выражение прямо в скобках.


```python
print(f'x = {2}')
```

`split` режет строку по пробелам и собирает список строк.


```python
print('10 20 20 30'.split())
```

`join` склеивает строки. Метод вызывают у строки-клея.


```python
print('-'.join(['q', 'w', 'e']))
```

`strip` убирает пробелы по краям.


```python
print('  qwerty  '.strip())
```

`replace` собирает новую строку, кусок в ней заменён.


```python
print('qwerty'.replace('we', 'WE'))
```

`find` даёт индекс куска. Если куска нет, результат `-1`.


```python
print('qwerty'.find('w'))
print('qwerty'.find('9'))
```

`count` считает, сколько раз встретился кусок.


```python
print('qwerty'.count('q'))
```

`lower` собирает новую строку в нижнем регистре.


```python
print('qwerty'.lower())
print('QWERTY'.lower())
```

`upper` собирает новую строку в верхнем регистре.


```python
print('qwerty'.upper())
```

`isdigit` проверяет, состоит ли строка целиком из цифр.


```python
print('qwerty'.isdigit())
print('10'.isdigit())
```

`startswith` проверяет начало строки.


```python
print('qwerty'.startswith('qw'))
print('qwerty'.startswith('a'))
```

`endswith` проверяет конец строки.


```python
print('qwerty'.endswith('ty'))
print('qwerty'.endswith('a'))
```

`int` от строки с пробелами останавливает программу: это не одно число.

`split` даёт строки. `map` применяет `int` к каждой. `list` собирает список чисел, `set` собирает множество.


```python
int('10 20 20 30')
```

```python
numbers = list(map(int, input().split()))
numbers
```

```python
set(map(int, input().split()))
```
