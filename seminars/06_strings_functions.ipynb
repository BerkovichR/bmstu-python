## Строки

Нумерация с нуля. Отрицательный номер считает с конца. Правый конец в срез не входит.


```python
s = 'qwerty'
print(s[0])
print(s[-1])
print(s[0:3])
```

Присваивание символа в строку останавливает программу.


```python
s[0] = 'a'
```

`str` переводит число в строку. `int` переводит строку из цифр в число.


```python
print(str(10))
print(int('10'))
print(type(str(10)), type(int('10')))
```

```python
int('10 20 20 30')
```

В f-строке фигурные скобки заменяются значением.


```python
x = 2
print(f'{x}')
```

`split` режет строку по пробелам и собирает список строк.


```python
print('10 20 20 30'.split())
```

## Меню

| Операция | Что делает |
|---|---|
| `s[i]` | символ по номеру с нуля, `s[-1]` с конца |
| `s[start:end]` | новая строка, правый конец не входит |
| `len(s)` | длина |
| `s + t` | новая строка, склейка |
| `s * n` | новая строка, повтор `n` раз |
| `str(x)` | число в строку |
| `int(s)` | строка из цифр в число |
| `f'{x}'` | подстановка значения в строку |
| `'{}'.format(x)` | то же через `format` |
| `s.split()` | список строк по пробелам |
| `s.strip()` | убрать пробелы по краям |
| `sep.join(parts)` | собрать строку из списка, между элементами `sep` |
| `s.replace(a, b)` | новая строка с заменой |
| `s.find(sub)` | номер первого вхождения или `-1` |
| `s.count(sub)` | сколько раз встретился фрагмент |
| `s.lower()`, `s.upper()` | новая строка в другом регистре |
| `s.isdigit()` | состоит ли строка из цифр |
| `s.startswith(prefix)` | начинается ли с фрагмента |
| `s.endswith(suffix)` | кончается ли фрагментом |

`len`, `str` и `int` вызывают как функции. Методы пишут через точку у строки.

## Функции

Один и тот же подсчёт для двух строк. Пока функции нет, цикл копируют целиком.


```python
text = 'мама мыла раму'
n = 0
for word in text.split():
    n = n + 1
print(n)

text = 'и всё'
n = 0
for word in text.split():
    n = n + 1
print(n)
```

Цикл написан один раз. Вторая строка считается вызовом, а не второй копией цикла.


```python
def word_count(text):
    n = 0
    for word in text.split():
        n = n + 1
    return n

print(word_count('мама мыла раму'))
print(word_count('и всё'))
```

Имя говорит, что функция делает. Тело `def` и тело `for` пишутся со следующей строки.


```python
def shout(s):
    result = ''
    for ch in s:
        result = result + ch + '!'
    return result

print(shout('ab'))
```

```python
def pow(x=3,n):
    return x * n
print(pow(2))
```

Без `return` вызов отдаёт `None`.


```python
def silent(s):
    result = s + s

print(silent('ab'))
```

Если аргумент не передали, подставляется значение по умолчанию.


```python
def pad(s, n=3):
    return s * n

print(pad('ab'))
```

Аргумент можно передать по имени.


```python
print(pad('ab', n=2))
```

`return` с запятой отдаёт кортеж.


```python
def min_max(a, b):
    if a < b:
        return a, b
    return b, a

print(min_max(3, 1))
lo, hi = min_max(3, 1)
print(lo, hi)
```

Первый параметр принимает один аргумент. `*parts` собирает остальные в кортеж.


```python
def glue(*parts, first):
    result = first
    for p in parts:
        result = result + p
    return first, parts, result

print(glue('a', 'b', 'c'))
```

```python
def report(name, *scores):
    return name, scores
```

Функция может прочитать внешнюю переменную.


```python
x = 2

def read_x():
    return x

print(read_x())
```

Присваивание внутри создаёт локальную переменную. Внешняя остаётся на месте.


```python
def write_local():
    x = 3
    return x

print(write_local())
print(x)
```

`global` пишет во внешнюю переменную.


```python
def set_global():
    global x
    x = 3

set_global()
print(x)
```

`nonlocal` пишет в переменную внешней функции.


```python
def outer():
    n = 0
    def inner():
        nonlocal n
        n = n + 1
        return n
    return inner() + inner()

print(outer())
```
