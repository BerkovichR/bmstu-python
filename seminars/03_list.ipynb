## Список

Переменная — это имя ссылки, не сам объект. Список — тоже объект. Его удобно видеть как **шапку** плюс массив ссылок.

Ниже схема объекта списка в CPython. На занятии разберём два поля. Остальное — общая картина: если хотите, изучайте сами.

```
PyListObject
┌──────────────────────────┐
│ PyObject_HEAD             │   служебное, как у любого объекта
│ ob_size                   │  ← количество элементов  (это len)
│ ob_item                   │  → массив PyObject*
│ allocated                 │  ← выделенная ёмкость
└──────────────────────────┘
```

- `ob_size` — длина, то что даёт `len(a)`;
- `ob_item` — указатель на массив ссылок.

Индекс — смещение в этом массиве. `a[0]` — нулевая ссылка, `a[2]` — вторая.

```
имя a  →  шапка (ob_size = 3)
              ob_item →  [ • ][ • ][ • ]
                           ↓    ↓    ↓
                           10   20   30
```

`PyObject_HEAD` и `allocated` на чертеже есть специально: шапка больше, чем «три числа подряд». Ёмкость — сколько ссылок влезет **без** нового массива.


### Создание, индекс, длина

Ячейки массива хранят **ссылки**, поэтому в одном списке могут лежать разные типы.

**Правила хорошего тона**

Тело `if` и `for` надо писать со **следующей** строки, с отступом. Не в ту же строку после двоеточия.


````python
a = [10, 20, 30]    # Создание списка
print(a)            # Печать списка
print(len(a))       # Длина списка
print(a[0])         # Первый элемент по индексу
print(a[2])         #
print(a[-1])        # Первый с конца
````

````python
a = [10, 20, 30]
a[1] = 99           # Поменять значение
print(a)
````

````python
a = [10, "qwerty", None]
print(a)
print(type(a[0]), type(a[1]))
````

Индекс вне диапазона — ошибка. Длина 3, законные индексы `0, 1, 2` и `-1, -2, -3`.

Контрольный вопрос: как очистить список?

## Методы списка

Метод вызывается через точку: `a.append(2)`. Он меняет **этот** список, а не создаёт новый.

| Метод            | Что делает                                       |
|------------------|--------------------------------------------------|
| `a.append(x)`    | добавить `x` в конец, длина +1                   |
| `a.pop()`        | снять и вернуть последний, длина −1              |
| `a.pop(i)`       | убрать элемент с индексом `i`, длина −1           |
| `a.insert(i, x)` | вставить `x` на место `i`, хвост сдвинуть вправо |
| `a.remove(x)`    | убрать первое вхождение значения `x`             |
| `a.sort()`       | упорядочить **этот** список по возрастанию       |



````python
a = [4, 1, 7]
a.append(2)
print("после append:", a)
last = a.pop()
print("pop вернул", last, "осталось", a)
a.insert(1, 9)
print("после insert:", a)
a.remove(4)
print("после remove:", a)
a.sort()
print("после sort:", a)
a.pop(1)
print("после pop:", a)
````

## Срезы

Запись `[start:end:step]`, или `[start:end]` если шаг `1`.

- `start` — откуда начинаем (по умолчанию `0`);
- `end` — докуда, **не включая** этот индекс (по умолчанию длина);
- `step` — шаг (по умолчанию `1`).

Поэтому `a[1:5]` и `a[1:5:1]` — одно и то же: шаг 1 можно не писать.

Срез строит **новый список**. Её ячейки смотрят на те же объекты, что и у исходного списка.

Отрицательный шаг: идти с конца.


````python
a = [10, 20, 30, 40, 50, 60]
print(a[1:5])
print(a[1:5:1])
print(a[1:5:2])
print(a[:3])
print(a[3:])
print(a[::2])
print(a[::-1])
````

## Цикл по списку

Два рабочих способа.

1. По значениям: `for x in a` — когда индекс не нужен.
2. По индексам: `for i in range(len(a))` — когда нужны соседи `a[i]` и `a[i + 1]`.

Тело цикла — со следующей строки, с отступом.


````python
a = [4, 1, 7, 2]
for x in a:
    print(x)
````

````python
a = [4, 1, 7, 2]
for i in range(len(a)):
    print(i, a[i])
````

### Максимум в списке

Дан список. Найти наибольший элемент.

Идея. Запоминаем кандидата (сначала `a[0]`). Идём по остальным. Если встретили больше — кандидат меняется.


````python
a = [4, 1, 7, 2]
m = a[0]
for x in a:
    if x > m:
        m = x
print(m)
````

````python
a = [4, 1, 7, 2]
m = a[0]
for i in range(len(a)):
    if a[i] > m:
        m = a[i]
print(m)
````

## Сортировка пузырьком

Задача. Переставить элементы так, чтобы каждый следующий был не меньше предыдущего: `a[0] <= a[1] <= ...`.

Идея. Сравниваем соседей. Если левый больше правого — меняем местами. Большой элемент «всплывает» вправо, как пузырёк в воде. Один проход по парам соседей ставит самый большой в конец. Повторяем проходы, пока не пройдём весь список.

Два вложенных цикла по длине $n$ — порядка $n^2$ сравнений.


### Схема

Цикл на схеме — ромб и стрелка назад, как на лекции.

```mermaid
flowchart TD
    A([начало]) --> B["a = [4, 1, 7, 2]"]
    B --> C["n = длина a"]
    C --> D["i = 0"]
    D --> E{"i < n"}
    E -->|нет| Z[/напечатать a/]
    Z --> END([конец])
    E -->|да| F["j = 0"]
    F --> G{"j < n - 1"}
    G -->|нет| I["i = i + 1"]
    I --> E
    G -->|да| H{"a[j] > a[j+1]"}
    H -->|да| S["обмен a[j] и a[j+1] через t"]
    H -->|нет| J["j = j + 1"]
    S --> J
    J --> G
```


### Шаги алгоритма

1. Дан список `a`. Запомнить длину `n`.
2. Положить `i = 0`.
3. Пока `i < n`:
    1. Положить `j = 0`.
    2. Пока `j < n - 1`:
        1. Если `a[j] > a[j + 1]`, обменять через `t`:
           `t = a[j]`, затем `a[j] = a[j + 1]`, затем `a[j + 1] = t`.
        2. Увеличить `j` на 1, вернуться к проверке `j < n - 1`.
    3. Увеличить `i` на 1, вернуться к проверке `i < n`.
4. Напечатать `a`.

Пример для списка `[4, 1, 7, 2]`:

| Сравниваем | Условие | Список после |
|---|---|---|
| 4 и 1 | 4 > 1, обмен | `[1, 4, 7, 2]` |
| 4 и 7 | 4 > 7? нет | `[1, 4, 7, 2]` |
| 7 и 2 | 7 > 2, обмен | `[1, 4, 2, 7]` |

Семёрка уехала в конец. Следующие проходы досортируют хвост слева.


Обмен на схеме — три действия через `t`. В Python тот же обмен можно записать одной строкой: `a[j], a[j + 1] = a[j + 1], a[j]`. Смысл тот же.


````python
a = [4, 1, 7, 2]
n = len(a)
for i in range(n):
    for j in range(n - 1):
        if a[j] > a[j + 1]:
            t = a[j]
            a[j] = a[j + 1]
            a[j + 1] = t
print(a)
````

Тот же смысл, обмен в одну строку. Контроль: должно получиться то же, что у `sort`.


````python
a = [4, 1, 7, 2]
n = len(a)
for i in range(n):
    for j in range(n - 1):
        if a[j] > a[j + 1]:
            a[j], a[j + 1] = a[j + 1], a[j]
print(a)
````

После первого прохода самый большой уже в конце. Внутренний цикл можно укоротить: `range(n - 1 - i)`. Это не новый алгоритм, та же схема с другим диапазоном `j`. Если на паре нет минут — ячейку не запускаем.


````python
a = [4, 1, 7, 2]
n = len(a)
for i in range(n):
    for j in range(n - 1 - i):
        if a[j] > a[j + 1]:
            t = a[j]
            a[j] = a[j + 1]
            a[j + 1] = t
print(a)
````

## Сортировка вставками

Второй метод сортировки. Описание ниже — чтобы можно было разобраться самостоятельно.

Идея. Левая часть списка уже упорядочена. Берём следующий элемент справа и **вставляем** его на своё место слева, сдвигая большие элементы вправо. Как карты в руке: новую карту вставляем между меньшими и большими.

Шаги.

1. Для `i` от 1 до `n - 1` (нулевой элемент — уже «левая часть» из одного элемента).
2. Запомнить `x = a[i]` — карту, которую вставляем.
3. Пока слева есть элементы больше `x`, сдвигать их на одну позицию вправо.
4. Поставить `x` в освободившееся место.
5. После последнего `i` весь список упорядочен.

│ ob_item                   │  → массив ссылок

| Метод | Что делает |

Срез: `a[start:end:step]`, шаг по умолчанию 1. Очистить шапку: `a.clear()`, не `a = []`.
