## Подходы к программированию

Новый подход появлялся, когда прежний переставал выдерживать размер программы. Ниже у каждой ступени указан условный масштаб проекта. Это не статистика и не норма. Это порядок величины, при котором программу ещё можно писать, читать и чинить.

### 1. Машинное программирование

1940–1950-е годы. Программа записывается машинным кодом: цепочкой двоичных инструкций процессора. Программист указывает, какое число положить в регистр, какую команду выполнить и к какой команде перейти дальше.

Сложить два числа — несколько команд. Прочитать первое число. Прочитать второе. Сложить. Записать результат. У этого сложения нет имени. Повторно его не вызвать: те же команды копируют ещё раз.

Затем появился ассемблер. У команды есть буквенное имя, но за именем по-прежнему одна команда процессора. Регистры и порядок команд программист выбирает сам.

Такой текст трудно писать: короткая задача занимает длинную ленту команд. Трудно читать через неделю: по ленте не видно, где кончается один смысл и начинается другой. Трудно искать ошибку: она прячется в одной команде длинной цепочки. Трудно переносить на другой процессор: там другие команды, и ленту пишут заново.

Условный масштаб: десятки — сотни строк.

### 2. Процедурное программирование

1950–1960-е годы. Языки Fortran и ALGOL. Появляются переменные, циклы, условия, функции и процедуры. Программист описывает алгоритм, а не команды процессора.

Повторяющийся кусок получает имя. Имя вызывают снова, с другими данными. Вызов заменяет новую копию тех же команд.

Функции, которые вы уже писали это процедурное программирование. У функции есть имя, параметры и результат. Один и тот же вызов годится для новых данных. Пока все имена лежат в одной программе, этого хватает.

Дальше программа становится слишком длинной. Одно и то же имя нужно двум программам, и функцию копируют во вторую. Копии живут отдельно. Починку вносят в каждую, и они расходятся.

Условный масштаб: сотни — тысячи строк.

### 3. Структурное и модульное программирование

1960–1970-е годы. Языки Pascal, C и Modula-2.

Здесь две разные идеи.

Структурное программирование описывает одну функцию. Действия внутри неё идут по порядку: ветвятся условием и повторяются циклом. Так вы уже пишете.

Модульное программирование описывает всю программу. Программу делят на части. У части небольшой интерфейс: имена, которые вызывают снаружи. Остальное спрятано внутри. Часть пишут и проверяют отдельно. Две программы вызывают одни и те же имена. Правка делается в одном месте.

Главная идея: разделить программу на независимые части и этим уменьшить сложность.

Условный масштаб: тысячи — десятки тысяч строк.

| Подход | Условный масштаб проекта |
|---|---|
| Машинное программирование | Десятки — сотни строк |
| Процедурное программирование | Сотни — тысячи строк |
| Модульное программирование | Тысячи — десятки тысяч строк |


Модуль в Python — файл с именами.


### 4. Объектно-ориентированное программирование

С конца 1960-х по 1990-е годы. Языки Simula, Smalltalk, C++ и Java. Данные и операции над ними объединяются в объект.

Идея: система — это объекты, которые вызывают операции друг друга. Условный масштаб поднимается ещё раз: десятки тысяч — сотни тысяч строк.

### 5. Аспектно-ориентированное программирование

Конец 1990-х годов. Аспектно-ориентированное программирование (АОП, AOP) — парадигма программирования, позволяющая выделять сквозную функциональность (логирование, безопасность, транзакции, обработку ошибок) в отдельные модули — аспекты, не смешивая её с основной бизнес-логикой приложения.

Основная идея: отделить функциональность, которая используется во многих частях программы, от функциональности, отвечающей за конкретные задачи этих частей.

### 6. Функциональное программирование

Это отдельная парадигма. Язык Lisp относится к 1958 году: функциональный стиль старше и объектного, и аспектного. Функция получает данные и возвращает результат. По дороге она не меняет переданные данные.

### 7. Декларативное и реактивное программирование

Декларативная программа описывает, какой результат нужен. Шаги, которыми результат получают, остаются за кадром.

Реактивная программа отвечает на события. Пришло событие — выполнилась реакция на него.

| Подход | Условный масштаб проекта          |
|---|-----------------------------------|
| Объектно-ориентированное программирование | Десятки тысяч — сотни тысяч строк |
| Аспектно-ориентированное программирование | Пока без ограничений              |

## Стандартная библиотека

Стандартная библиотека — модули, которые ставятся вместе с Python. Их не кладут в папку проекта. Имя забирают через `import`, как у своего файла.

`math` уже открыт. Это математика стандартной библиотеки.

`random` генерирует псевдослучайное число. При каждом запуске программы результат может отличаться. `seed` задаёт начальное состояние генератора, благодаря чему при одинаковом начальном значении последовательность чисел будет воспроизводиться.

`datetime` хранит дату. Разность двух дат показывает, сколько дней между ними.

`os` спрашивает операционную систему о файлах.

`sys` описывает запущенный Python. Список `sys.path` — папки, в которых `import` ищет модули. На экране будет первая папка этого списка. Пустая строка означает текущую папку. Длинный путь означает папку ноутбука. В обоих случаях поиск начинается рядом с ноутбуком, поэтому `import mymath` находит файл, который лежит здесь.


````python
import math
# наибольший общий делитель
print(math.gcd(12, 18))
````

````python
import random
random.seed(1) # задаёт начальное состояние генератора
print(random.randint(1, 6))
````

````python
from datetime import date
start = date(2026, 9, 1)
day = date(2026, 11, 8)
print((day - start).days)
````

````python
import os
# есть ли файл?
print(os.path.isfile("mymath.py"))
````

````python
import sys
print(sys.path[0])
````

## Пакет

Модуль — один файл с именами. Пакет — папка таких файлов. В папке лежит `__init__.py`. По этому файлу Python отличает пакет от обычной папки.

Пакет `geom` собран так:

```
geom/
    __init__.py
    points.py
```

`points.py` — модуль. В нём функция `distance`: расстояние между двумя точками. Строка `import math` внутри файла забирает имя из стандартной библиотеки так же, как это делает программа.

Пустой `__init__.py` тоже делает папку пакетом. Тогда имя берут так: `geom.points.distance`. В нашем файле одна строка. Она поднимает `distance` из модуля `points` на уровень пакета, и имя видно как `geom.distance`.

```python
from geom.points import distance
```

`import geom` даёт пакет. `geom.distance` — имя из `__init__.py`. `geom.points.distance` — то же имя из файла `points.py`.

Свой пакет собирают тем же шагом, что и `mymath.py`. Функцию пишут в файл модуля. Файл кладут в папку. Рядом добавляют `__init__.py`. Программа забирает имя через `import`.


````python
import geom
print(geom.distance(0, 0, 3, 4))
print(geom.points.distance(0, 0, 3, 4))
````
