# Семинар 11. NumPy, одномерный массив


## Массив `ndarray`

Основной объект библиотеки NumPy — массив `ndarray`.

У него три свойства.

Первое — однородность. Все элементы одного типа. Имя типа — `dtype`. Элементы лежат в памяти одним сплошным рядом, и у каждого один и тот же размер. Место элемента с номером `i` получают так: к началу ряда прибавляют `i`, умноженный на размер элемента. Формула: `base + i * size`. Одинаковый размер и даёт однородность. Если бы соседние элементы имели разный размер, шаг от одного к другому менялся бы, и по номеру элемент было бы не найти.

Второе — многомерность. Такой ряд можно читать как вектор: у него одно измерение. Тот же способ хранения годится для матрицы, у неё два измерения, строки и столбцы, и для массива с большим числом измерений. Сегодня вектор, таблица в следующий раз.

Третье — эффективность. Действие над массивом записано без цикла в тексте программы. Одно выражение считает все элементы. Поэлементный проход выполняет библиотека.

Такой массив принимают другие вычислительные библиотеки.


```python
import numpy as np
```

## Список и массив

Плюс склеивает списки, звёздочка повторяет список.


```python
a = [1, 2, 3]
b = [4, 5, 6]
print(a + b)
print(a * 2)
```

Плюс складывает элементы, стоящие на одном месте. Звёздочка умножает каждый элемент на число.


```python
na = np.array([1, 2, 3])
nb = np.array([4, 5, 6])
print(na + nb)
print(na * 2)
```

Список целиком в квадрат не возводится.


```python
print([1, 2, 3] ** 2)
```

Массив возводит в квадрат каждый элемент.


```python
print(np.array([1, 2, 3]) ** 2)
```

Цикл собрал новый список.


```python
a_list = [1, 2, 3]
b_list = [0, 2, 1]
c = []
for i in range(len(a_list)):
    c.append((a_list[i] - b_list[i]) ** 2)
print(c)
```

Массив считает ту же формулу сразу по всем местам.


```python
a = np.array([1, 2, 3])
b = np.array([0, 2, 1])
print((a - b) ** 2)
```

Деление массива на число даёт дроби.


```python
print(np.array([1, 2, 3]) / 2)
```

## Тип элементов

Элемент с номером `i` лежит по адресу `base + i * size`.

В списке эти три числа так и лежат рядом. Массив выбирает один тип на всех. Дробь среди целых поднимает весь массив до `float64`.


```python
x = np.array([1, 2, 3.5])
print(x, x.dtype)
```

Дробная часть пропала без сообщения об ошибке. В ячейке целого типа для `2.9` места нет.


```python
a = np.array([1, 2, 3])
a[0] = 2.9
print(a, a.dtype)
```

Список такое значение хранит.


```python
s = [1, 2, 3]
s[0] = "x"
print(s)
```

В целый массив строка не встаёт.


```python
a = np.array([1, 2, 3])
a[0] = "x"
```

## `arange` и `zeros`

Правило края то же, что у `range`. Число 6 в ответ не входит.


```python
print(list(range(0, 6, 2)))
print(np.arange(0, 6, 2))
```

`zeros` сразу кладёт элементы одного типа. Без `dtype` это дроби. Целые нули просят явно.


```python
print([0] * 3)
z = np.zeros(3)
print(z, z.dtype)
zi = np.zeros(3, dtype=int)
print(zi, zi.dtype)
```

Разные способы создания массивов

```python
# Массив из нулей
np.zeros(5)

# Массив из единиц
np.ones((2, 3))

# Массив из одинаковых значений
np.full(5, 10)

# Последовательность чисел
np.arange(0, 10, 2)  # [0, 2, 4, 6, 8]

# Равномерно распределённые числа
np.linspace(0, 1, 5)  # [0., 0.25, 0.5, 0.75, 1.]
```

## Атрибуты

Запятая в `(6,)` держит кортеж из одного числа. `ndim` считает измерения, здесь оно одно. `size` считает элементы. `dtype` — тип элементов.


```python
a = np.arange(6)

print(a.shape) # Размерность: (6,)
print(a.ndim) # Количество измерений: 1
print(a.size)  # Количество элементов: 6
print(a.dtype)  # Тип элементов int64
print(a.itemsize) # Размер одного элемента в байтах
```

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
print(a.shape) # Размерность: (2,3)
print(a.ndim) # Количество измерений: 2
print(a.size)  # Количество элементов: 6
print(a.dtype)  # Тип элементов int64
print(a.itemsize) # Размер одного элемента в байтах
```

## Математические функции
Большинство математических функций NumPy работают сразу со всем массивом.

```python
a = np.array([1, 4, 9, 16])

np.sqrt(a)    # Квадратный корень
np.abs(a)     # Модуль
np.exp(a)     # Экспонента
np.log(a)     # Натуральный логарифм
np.sin(a)     # Синус
np.cos(a)     # Косинус
np.round(a)   # Округление
```

```python
np.min(a)
np.max(a)
np.sum(a)
np.mean(a)  # Среднее
np.median(a) # Медиана
np.std(a)   # Стандартное отклонение
np.var(a)   # Дисперсия
```

Логические операции и фильтрация

```python
a = np.array([1, 5, 8, 3, 10, 2])

# & (И)
# | (ИЛИ)
# ~ (НЕ)
print(a > 5)

print(a[a > 5])
```

## Индекс, срез, вид

Индекс и срез пишутся так же, как у списка.


```python
a = np.arange(6)
print(a[0], a[-1])
print(a[1:4])
```

`b` смотрит на те же элементы, что и кусок `a`. Запись в `b` меняет `a`. У списка срез уносит копию, и исходный список от такой записи не меняется.


```python
a = np.arange(6)
b = a[1:3]
b[0] = 99
print(b)
print(a)
```

`.copy()` забирает элементы в новый массив. Исходный ряд остаётся.


```python
a = np.arange(6)
c = a[1:3].copy()
c[0] = 99
print(c)
print(a)
```

```python
import numpy as np
import time

n = 10_000_000

a = list(range(n))
b = list(range(n))

# Ужасно!
start = time.perf_counter()
c = []
for i in range(n):
    c.append(a[i]+b[i])
print("For:", time.perf_counter() - start)

# Уже лучше.
start = time.perf_counter()
c = [a[i] + b[i] for i in range(n)]
print("List:", time.perf_counter() - start)

a = np.arange(n)
b = np.arange(n)

# numpy.
start = time.perf_counter()
c = a + b
print("NumPy:", time.perf_counter() - start)
```
