## Команды ввода и вывода данных

Вывести на экран результаты работы программы можно с помощью функции print.
В наиболее общем виде функция имеет следующий формат:
`print(objects, sep=' ', end='\n', file=sys.stdout, flush=False)`
В этой функции
- objects – один или несколько объектов, которые нужно вывести;
- sep – разделитель между объектами (по умолчанию пробел);
- end – символ завершения строки (по умолчанию перевод строки);
- file – объект файла, в который направлен вывод (по умолчанию на экран);
- flush – принудительный сброс буфера вывода (по умолчанию сброса нет).

Любой из этих аргументов при вызове функции можно опустить.
Тогда для отсутствующих аргументов будет использоваться значение по умолчанию.
Аргумент objects в функции – это то, что мы хотим вывести на экран.
Выводить можно строки, числа, переменные, выражения и т.д.
После вывода objects функция print по умолчанию добавляет символ перевода строки \n.
Он на экране не отображается, но его присутствие указывает интерпретатору, что в этом месте надо перейти на новую строку.

````python
print('Строка 1')
print('Строка 2')
````

Если необходимо вывести результат на той же строке, то в функции print данные указываются через запятую.

````python
print('Строка1', 'Строка2')
````

Между выводимыми строками по умолчанию вставляется пробел.
С помощью параметра sep можно указать другой символ или последовательность символов.
Например, запятую и пробел:

````python
print('Строка1', 'Строка2', sep=', ')
````

Как было сказано ранее, после вывода объектов в конце добавляется символ перевода строки.
Если необходимо произвести вывод при следующем вызове функции print на той же строке, то в параметре end следует указать другой символ или другую последовательность символов.
Например, запятую и пробел:

````python
print('Строка1', 'Строка2', sep=', ', end=', ')
print('Строка3')
````

Для вывода содержимого переменной достаточно указать ее имя в качестве аргумента функции print:

````python
name='Алена'
print('Меня зовут', name)
````

Пример вывода числа и выражения:

````python
print(3, 4+5)
````

Для ввода данных предназначена функция input. Она имеет следующий формат:

`переменная=input(сообщение)`

Интерпретатор выводит на экран сообщение и переходит в режим ожидания
ввода. После ввода пользователем данных и нажатия клавиши Enter
интерпретатор записывает результат ввода в переменную. Переменная и
сообщение при вызове функции могут быть опущены.

````python
name=input('Введите ваше имя:')
print('Привет,',name, '!')
input('Нажмите <Enter> для завершения...')
````

Функция input всегда возвращает данные в виде строки, даже если пользователь вводит число. Для получения числовых значений необходимо использовать функции преобразования типов данных:
- int(значение) – преобразует значение в целое число,
- float(значение) – преобразует значение в вещественное число.

Например, следующая программа запрашивает у пользователя произвольное целое число, увеличивает его на 2 и выводит на экран:

````python
number=int(input('Введите целое число: '))
print(number+2)
````

В данной программе делается явное преобразование типов. Функция `input` возвращает введенное пользователем целое число в виде строки, а функция `int` приводит эту строку к целочисленному типу.
В примере пользователь вводит число 3. Функция input возвращает строку '3' и передает ее в качестве входного аргумента функции `int`. Далее функция `int` переводит строку '3' в число 3, после чего это число присваивается переменной `number`.
Если в данной программе явно не сделать преобразование типов

```
int(input(...))
```

то в переменную number запишется строка и при попытке выполнения операции сложения number+2 программа остановится и выдаст ошибку, поскольку операция сложения строки и числа в Python неопределенна.

## Форматированный вывод данных

Форматирование строк – это процесс создания новых строк по заданному шаблону, путем подстановки в него значений переменных или выражений.
Форматировать строки можно с помощью оператора `%`.
Он задает место в строке, куда должно быть подставлено значение переменной, и определяет, как будет выглядеть выводимое значение на экране в зависимости от типа переменной.
Например, выражение `%s` указывает, что в этом месте надо вставить строку, `%d` означает целое число, а `%f` – число с плавающей запятой.
Такие выражения называются спецификаторами формата, поскольку они определяют, как именно нужно интерпретировать и отображать значение переменной при выводе на
экран.

Приведем еще пару примеров.
Спецификатор `%.2f` говорить, что число с плавающей запятой при выводе на экран необходимо округлить до двух знаков после запятой.
Спецификатор %5d указывает, что минимальное количество цифр в выводимом целом числе должно быть 5.
Если цифр по факту меньше, то недостающие цифры при выводе заменяются пробелами слева от числа.

Оператор `%` считается классическим способом форматирования строк, но в современном Python уступает место f-строкам.
Приведем примеры использования оператора `%`:

````python
# Вывод значения одной переменной
name = 'Роман'
print('Привет, %s!' %name)
````

Обратите внимание, что переменная name пишется сразу после форматируемой строки и предваряется символом `%`, при этом конструкция %name отделяется от строки только пробелом.

````python
# Вывод значений нескольких переменных
age = 20
gpa = 4.75

print("Студент %s, возраст: %d, средний балл: %.1f" %(name, age, gpa))
````

Спецификатор `%.1f` в примере округляет выводимое число с плавающей точкой до одного знака после запятой – переменная gpa содержит значение `4.75`, которое выводится как `4.8`.

Все переменные name, age и gpa перечисляются после самой форматируемой строки в порядке вставки их значений, т.е. первый спецификатор `%s` соответствует первой переменной name и т.д.
Переменные обязательно должны быть выделены круглыми скобками и предваряться символом `%`.
